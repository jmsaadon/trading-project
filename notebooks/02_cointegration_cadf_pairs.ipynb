# Section 10.3 CADF Cointegration Walkthrough

This notebook mirrors the textbook AREX/WLL workflow for primary ETF pairs: price-series plot, scatter plot, no-intercept hedge-ratio residuals, and an ADF test on residuals. It is research diagnostics only, not a backtest.

```
import matplotlib.pyplot as plt
import pandas as pd

from trading_project.config import load_config
from trading_project.data.db import connect
from trading_project.data.retrieval import get_adjusted_close_matrix
from trading_project.research.cointegration import (
    PrimaryPair,
    align_pair_prices,
    plot_price_series,
    plot_residuals,
    plot_scatter_series,
    primary_pairs,
    run_cadf_test,
    summarize_cadf_tests,
)
from trading_project.research.descriptive import unique_relationship_tickers
from trading_project.research.relationships import load_relationships

pd.set_option('display.max_columns', 60)
plt.style.use('seaborn-v0_8-whitegrid')
```

```
config = load_config()
conn = connect(config.database_url)
relationships = load_relationships()
tickers = unique_relationship_tickers(relationships)
prices = get_adjusted_close_matrix(conn, tickers=tickers)
pairs = primary_pairs(relationships)
```

## CADF Summary Ranking

```
summary = summarize_cadf_tests(prices, relationships)
summary
```

## gdx_gld_spy: GDX / GLD

- Overlap: `2006-05-22` through `2026-05-19`
- Observations: `5030`
- No-intercept hedge ratio: `0.208268`
- ADF statistic on residuals: `-1.717550`
- p-value: `0.422093`
- 5% critical value: `-2.862118`
- Conclusion: **Fail to reject no-cointegration null**

![gdx_gld_spy CADF workflow](../reports/figures/cointegration/gdx_gld_spy_cadf.png)

```
pair = PrimaryPair(relationship='gdx_gld_spy', target='GDX', explanatory='GLD')
pair_prices = align_pair_prices(prices, pair.target, pair.explanatory)
result, residuals = run_cadf_test(pair, pair_prices)
print(result)
fig, axes = plt.subplots(3, 1, figsize=(11, 10))
plot_price_series(pair_prices, pair.target, pair.explanatory, ax=axes[0])
plot_scatter_series(pair_prices, pair.target, pair.explanatory, ax=axes[1])
plot_residuals(residuals, pair.relationship, ax=axes[2])
fig.tight_layout()
```
